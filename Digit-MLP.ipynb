Model Description 
This ML model Code implements a TensorFlow/Keras deep learning pipeline designed to classify handwritten digits (0–9) using the canonical MNIST dataset, complete with Training and Testing loss graphs, Confusion matrix and output images.

This Model uses MLP 
Multi-Layer Perceptron (MLP): 
Tried making the model using MLP, where The core script uses dense, fully connected layers that flatten the 28x28 image into a 1D array of 784 pixels. It relies on fixed pixel coordinates, making it sensitive to shifts, scaling, or distortion in real-world handwriting.
The computing speed was faster (around 3-4 secs per epoch) and the accuracy of testing and training were in the range of 98%. But the issue is that the accuracy of the model with user uploaded image was a bit low.

Dataset and Data Split
•	Training Set: 60,000 images and labels (x_train, y_train) used for backpropagation and weight optimization.
•	Testing Set: 10,000 independent images and labels (x_test, y_test) reserved strictly for final, unbiased model evaluation.

Epochs and Their Function
 An epoch is a single complete cycle where the model processes the entire training dataset. Setting epochs=15 gives the network multiple opportunities to refine its parameters, while the Early Stopping callback (patience=3) monitors validation loss and halts training automatically if performance plateaus, preventing overfitting and saving time.


Batch size : 100   -> when it is increased, then the proccessing speed is more but the accuracy drops down
Epochs : 15 -> When it is increased, the system gets more iteration to train.
Layers : 3 trainable, 2 hidden -> Adding layers quickly causes massive parameter bloat, severe overfitting, and vanishing gradients, which rapidly degrades performance.
Learning rate : 0.001  -> A high learning rate causes the network to wildly overshoot optimal weights and fail to converge, while a low learning rate makes training agonizingly slow and prone to getting stuck in sub-optimal solutions.



```python
import io
from PIL import Image, ImageOps
import ipywidgets as widgets
from IPython.display import display, clear_output

import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.datasets import mnist

from sklearn.metrics import confusion_matrix, classification_report

print("TensorFlow version:", tf.__version__)
print("NumPy version:", np.__version__)
print("GPU devices:", tf.config.list_physical_devices("GPU"))


# ============================================================
# 2. LOAD MNIST DATASET
# ============================================================

(x_train, y_train), (x_test, y_test) = mnist.load_data()

print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", x_test.shape)
print("Testing labels:", y_test.shape)


# ============================================================
# 3. DISPLAY SAMPLE IMAGES
# ============================================================

plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 4. DISPLAY ONE IMAGE
# ============================================================

plt.figure(figsize=(6, 6))
plt.imshow(x_train[0], cmap="gray")
plt.colorbar()
plt.title(f"Digit: {y_train[0]}")
plt.show()


# ============================================================
# 5. CHECK ORIGINAL PIXEL VALUES
# ============================================================

print("Image shape:", x_train[0].shape)
print("Minimum pixel value:", x_train[0].min())
print("Maximum pixel value:", x_train[0].max())


# ============================================================
# 6. NORMALIZE PIXEL VALUES
# ============================================================

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Normalized minimum:", x_train.min())
print("Normalized maximum:", x_train.max())

# ============================================================
# 7. BUILD FAST MLP MODEL (Replaces CNN)
# ============================================================
# This deeper Dense architecture is extremely fast to train 
# and comfortably exceeds 98% accuracy on MNIST.

model = keras.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    
    layers.Dense(512, activation="relu"),
    layers.Dropout(0.2),
    
    layers.Dense(256, activation="relu"),
    layers.Dropout(0.2),
    
    layers.Dense(10, activation="softmax")
])

# ============================================================
# 8. DISPLAY MODEL ARCHITECTURE
# ============================================================

model.summary()


# ============================================================
# 9. COMPILE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 10. EARLY STOPPING
# ============================================================

early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

# ============================================================
# 10.5 CUSTOM CALLBACK FOR TEST LOSS
# ============================================================

class TestEvaluationCallback(keras.callbacks.Callback):
    def __init__(self, test_data):
        self.x_test, self.y_test = test_data
        self.test_losses = []
        
    def on_epoch_end(self, epoch, logs=None):
        loss, acc = self.model.evaluate(self.x_test, self.y_test, verbose=0)
        self.test_losses.append(loss)

test_tracker = TestEvaluationCallback((x_test, y_test))

# ============================================================
# 11. TRAIN MODEL
# ============================================================

history = model.fit(
    x_train,
    y_train,
    epochs=15,
    batch_size=100,             
    validation_split=0.1,  # Keeps the 54,000 / 6,000 split
    callbacks=[early_stopping, test_tracker],
    verbose=1
)


# ============================================================
# 12. DISPLAY TRAINING HISTORY
# ============================================================

print(history.history.keys())


# ============================================================
# 13. PLOT TRAINING LOSS VS EPOCH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss",
    color="blue"
)

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# 14. PLOT TEST LOSS VS EPOCH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    test_tracker.test_losses,
    label="Test Loss",
    color="orange"
)

plt.title("Test Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# 15. EVALUATE TRAINING DATA
# ============================================================

train_loss, train_accuracy = model.evaluate(
    x_train,
    y_train,
    verbose=0
)

print("=" * 45)
print("TRAINING PERFORMANCE")
print("=" * 45)
print(f"Training Accuracy: {train_accuracy * 100:.2f}%")
print(f"Training Loss:     {train_loss:.4f}")


# ============================================================
# 16. EVALUATE TEST DATA
# ============================================================

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("=" * 45)
print("TEST PERFORMANCE")
print("=" * 45)
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
print(f"Test Loss:     {test_loss:.4f}")
print("=" * 45)


# ============================================================
# 17. CHECK 98% REQUIREMENT
# ============================================================

if test_accuracy >= 0.98:
    print("✓ Test accuracy requirement of 98% is satisfied.")
else:
    print("✗ Test accuracy requirement of 98% is NOT satisfied.")


# ============================================================
# 18. GENERATE TEST PREDICTIONS
# ============================================================

y_pred_probability = model.predict(
    x_test,
    verbose=0
)

y_pred = np.argmax(
    y_pred_probability,
    axis=1
)


# ============================================================
# 19. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted Digit")
plt.ylabel("Actual Digit")
plt.title("MNIST Confusion Matrix")

plt.show()


# ============================================================
# 20. CLASSIFICATION REPORT
# ============================================================

print(
    classification_report(
        y_test,
        y_pred,
        digits=4
    )
)


# ============================================================
# 21. DISPLAY TEST PREDICTIONS
# ============================================================

plt.figure(figsize=(12, 6))

for i in range(15):

    plt.subplot(3, 5, i + 1)

    plt.imshow(
        x_test[i],
        cmap="gray"
    )

    actual = y_test[i]
    predicted = y_pred[i]

    plt.title(
        f"Actual: {actual} | Pred: {predicted}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 22. FIND INCORRECT PREDICTIONS
# ============================================================

incorrect_indices = np.where(
    y_pred != y_test
)[0]

print(
    "Number of incorrect predictions:",
    len(incorrect_indices)
)


# ============================================================
# 23. DISPLAY INCORRECT PREDICTIONS
# ============================================================

plt.figure(figsize=(12, 6))

for i, index in enumerate(incorrect_indices[:15]):

    plt.subplot(3, 5, i + 1)

    plt.imshow(
        x_test[index],
        cmap="gray"
    )

    plt.title(
        f"Actual: {y_test[index]}\n"
        f"Pred: {y_pred[index]}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 24. SAVE TRAINED MODEL
# ============================================================

model_path = "mnist_digit_model.keras"

model.save(model_path)

print("Model saved successfully!")
print("Model path:", os.path.abspath(model_path))
print("Model exists:", os.path.exists(model_path))


# ============================================================
# 25. DELETE MODEL FROM MEMORY
# ============================================================

del model

print("Original model removed from memory.")


# ============================================================
# 26. RELOAD SAVED MODEL
# ============================================================

loaded_model = keras.models.load_model(
    model_path
)

print("Saved model loaded successfully!")


# ============================================================
# 27. DISPLAY RELOADED MODEL SUMMARY
# ============================================================

loaded_model.summary()


# ============================================================
# 28. EVALUATE RELOADED MODEL
# ============================================================

reloaded_test_loss, reloaded_test_accuracy = loaded_model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("=" * 45)
print("RELOADED MODEL PERFORMANCE")
print("=" * 45)
print(
    f"Reloaded Test Accuracy: "
    f"{reloaded_test_accuracy * 100:.2f}%"
)

print(
    f"Reloaded Test Loss: "
    f"{reloaded_test_loss:.4f}"
)


# ============================================================
# 29. PREDICT USING RELOADED MODEL
# ============================================================

loaded_predictions = loaded_model.predict(
    x_test[:10],
    verbose=0
)

loaded_predicted_digits = np.argmax(
    loaded_predictions,
    axis=1
)

print("Actual labels:   ", y_test[:10])
print("Predicted labels:", loaded_predicted_digits)


# ============================================================
# 30. DISPLAY PREDICTIONS FROM RELOADED MODEL
# ============================================================

plt.figure(figsize=(12, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        x_test[i],
        cmap="gray"
    )

    actual = y_test[i]
    predicted = loaded_predicted_digits[i]

    plt.title(
        f"Actual: {actual}\n"
        f"Predicted: {predicted}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 31. VERIFY RELOADED MODEL
# ============================================================

if abs(test_accuracy - reloaded_test_accuracy) < 0.001:
    print("✓ Saved model successfully verified.")
else:
    print("⚠ Accuracy difference detected.")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import io
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1. Initialize widgets safely
uploader = widgets.FileUpload(accept='image/*', multiple=False)
predict_button = widgets.Button(description="Predict Digit", button_style='success')
output_area = widgets.Output()

def handle_prediction(b):
    with output_area:
        clear_output(wait=True)
        
        # Check if a file has been selected
        if not uploader.value:
            print("Please upload an image file first.")
            return
            
        try:
            # Handle ipywidgets v8 data structure safely
            uploaded_file = uploader.value[0]
            if isinstance(uploaded_file, dict):
                image_bytes = uploaded_file['content']
            else:
                image_bytes = uploaded_file.content
        except Exception:
            try:
                image_bytes = list(uploader.value.values())[0]['content']
            except Exception as e:
                print(f"Could not read uploaded file: {e}")
                return
                
        # Convert memoryview to raw bytes if necessary
        if hasattr(image_bytes, 'tobytes'):
            image_bytes = image_bytes.tobytes()
            
        # Load image into PIL
        try:
            original_img = Image.open(io.BytesIO(image_bytes))
            img_gray = original_img.convert('L')
        except Exception as e:
            print(f"Error opening image format: {e}")
            return

        # Preprocess for the model
        img_resized = img_gray.resize((28, 28), Image.Resampling.BILINEAR)
        img_array = np.array(img_resized).astype("float32")
        
        img_array = img_array - img_array.min()
        if img_array.max() > 0:
            img_array = img_array / img_array.max()
            
        if img_array.mean() > 0.4:
            img_array = 1.0 - img_array
            
        # Reshape to (1, 28, 28) to match the MLP's expected input shape
        img_batch = img_array.reshape(1, 28, 28)
        
        # Predict using loaded model or active model session
        target_model = loaded_model if 'loaded_model' in globals() else model
        probabilities = target_model.predict(img_batch, verbose=0)
        predicted_digit = np.argmax(probabilities[0])
        confidence = probabilities[0][predicted_digit]
        
        # Display side-by-side results
        fig, axes = plt.subplots(1, 2, figsize=(8, 4))
        
        axes[0].imshow(original_img)
        axes[0].set_title("Uploaded Input Image")
        axes[0].axis("off")
        
        axes[1].imshow(img_array, cmap="gray")
        axes[1].set_title(f"Model View\nPredicted: {predicted_digit} ({confidence * 100:.1f}%)")
        axes[1].axis("off")
        
        plt.tight_layout()
        plt.show()

# Bind click action to button
predict_button.on_click(handle_prediction)

# Render widgets neatly stacked
display(widgets.VBox([uploader, predict_button, output_area]))
```
